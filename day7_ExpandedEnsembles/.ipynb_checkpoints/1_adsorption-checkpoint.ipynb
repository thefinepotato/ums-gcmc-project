```python
from pymatgen.core import Lattice, Structure
import nglview

a, b, c = 26.343, 26.343, 26.343
alpha, beta, gamma = 90, 90, 90
lattice = Lattice.from_parameters(a, b, c, alpha, beta, gamma)
species = ["Cu", "O", "C", "C", "C", "H"]
coords = [
  [0.2853, 0.2853, 0],
  [0.3166, 0.2431, 0.9478],
  [0.2968, 0.2032, 0.9313],
  [0.322,  0.178,  0.887],
  [0.3655, 0.1994, 0.8655],
  [0.3802, 0.228,  0.8802]
]

structure = Structure.from_spacegroup(225, lattice, species, coords)
view = nglview.show_pymatgen(structure, default_representation=False)
view.add_unitcell()
view.add_spacefill()
view
```

```python
from raspalib import int3, double3, random, PseudoAtom, VDWParameters, Atom, ForceField, SimulationBox, Framework, MCMoveProbabilitiesParticles, Component, MCMoveProbabilitiesSystem, System, MonteCarlo
import sys
import numpy as np
import matplotlib.pyplot as plt
from tqdm.notebook import trange, tqdm

atomTypes = [PseudoAtom(name = "Cu1",   frameworkType = True,  mass = 63.5460, charge =  1.248,  atomicNumber = 29),
             PseudoAtom(name = "O1",    frameworkType = True,  mass = 15.9994, charge = -0.624,  atomicNumber =  8),
             PseudoAtom(name = "C1",    frameworkType = True,  mass = 12.0107, charge =  0.494,  atomicNumber =  6),
             PseudoAtom(name = "C2",    frameworkType = True,  mass = 12.0107, charge =  0.13,   atomicNumber =  6),
             PseudoAtom(name = "C3",    frameworkType = True,  mass = 12.0107, charge = -0.156,  atomicNumber =  6),
             PseudoAtom(name = "H1",    frameworkType = True,  mass = 1.00794, charge =  0.156,  atomicNumber =  6),
             PseudoAtom(name = "CH4",   frameworkType = False, mass = 16.0425, charge =  0.0,    atomicNumber =  6),
             PseudoAtom(name = "C_co2", frameworkType = False, mass = 12.0,    charge =  0.6512, atomicNumber =  6),
             PseudoAtom(name = "O_co2", frameworkType = False, mass = 15.9994, charge = -0.3256, atomicNumber =  8)]


parameters = [VDWParameters(2.5161, 3.11369),
              VDWParameters(48.1581, 3.03315),
              VDWParameters(47.8562, 3.47299),
              VDWParameters(47.8562, 3.47299),
              VDWParameters(47.8562, 3.47299),
              VDWParameters(7.64893, 2.84642),
              VDWParameters(158.5, 3.72),
              VDWParameters(29.933, 2.745),
              VDWParameters(85.671, 3.017)]

force_field = ForceField(pseudoAtoms = atomTypes,
                         parameters = parameters,
                         mixingRule = ForceField.MixingRule.Lorentz_Berthelot,
                         cutOffFrameworkVDW = 12.0,
                         cutOffMoleculeVDW = 12.0,
                         cutOffCoulomb = 12.0,
                         shifted = True,
                         tailCorrections = False,
                         useCharge = True)


framework = Framework(frameworkId = 0,
                      forceField = force_field,
                      componentName = "Cu-BTC",
                      simulationBox = SimulationBox(26.343, 26.343, 26.343),
                      spaceGroupHallNumber = 523,
                      definedAtoms = [
                        Atom(double3(0.2853, 0.2853, 0),       1.248, 1.0, 0, 0, 0, 0),
                        Atom(double3(0.3166, 0.2431, 0.9478), -0.624, 1.0, 0, 1, 0, 0),
                        Atom(double3(0.2968, 0.2032, 0.9313),  0.494, 1.0, 0, 2, 0, 0),
                        Atom(double3(0.322,  0.178,  0.887),   0.130, 1.0, 0, 3, 0, 0),
                        Atom(double3(0.3655, 0.1994, 0.8655), -0.156, 1.0, 0, 4, 0, 0),
                        Atom(double3(0.3802, 0.228,  0.8802),  0.156, 1.0, 0, 5, 0, 0)
                      ],
                      numberOfUnitCells = int3(1, 1, 1))

move_probabilities = MCMoveProbabilitiesParticles(translationProbability = 1.0,
                                                  rotationProbability = 1.0,
                                                  swapProbability = 1.0,
                                                  swapCBMCProbability = 0.0,
                                                  swapCFCMCProbability = 0.0,
                                                  swapCBCFCMCProbability = 0.0,
                                                  widomProbability = 0.0)

component = Component(componentId = 0,
                      forceField = force_field,
                      componentName = "CO2",
                      criticalTemperature = 304.1282,
                      criticalPressure = 7377300.0,
                      acentricFactor= 0.22394,
                      definedAtoms = [
                         Atom(double3(0.0, 0.0,  1.149), -0.3256, 1.0, 0, 7, 0, 0),
                         Atom(double3(0.0, 0.0,  0.0  ),  0.6512, 1.0, 0, 8, 0, 0),
                         Atom(double3(0.0, 0.0, -1.149), -0.3256, 1.0, 0, 7, 0, 0)
                      ],
                      numberOfBlocks = 5,
                      numberOfLambdaBins = 21,
                      particleProbabilities = move_probabilities,
                      fugacityCoefficient = 1.0,
                      thermodynamicIntegration = True)

system_probabilities = MCMoveProbabilitiesSystem(0.0, 0.0, 0.0, 0.0)

system = System(systemId = 0,
                forceField = force_field,
                simulationBox = None,
                temperature = 300.0,
                pressure = 1e4,
                heliumVoidFraction = 0.774,
                frameworkComponents = [ framework ],
                components = [ component ],
                initialNumberOfMolecules = [0],
                numberOfBlocks = 5,
                systemProbabilities = system_probabilities,
                sampleMoviesEvery = None)

#print(system)

seed = random(seed = 100)

mc = MonteCarlo(numberOfCycles = 5000,
                numberOfInitializationCycles = 5000,
                numberOfEquilibrationCycles = 5000,
                printEvery = 1000,
                writeBinaryRestartEvery = 5000,
                rescaleWangLandauEvery = 5000,
                optimizeMCMovesEvery = 5000,
                systems = [system],
                randomSeed = seed,
                numberOfBlocks = 5)

#print(mc)

progress = tqdm(total=3)
sys.stdout.flush()

mc.initialize()
progress.update(1)
sys.stdout.flush()

mc.equilibrate()
progress.update(1)
sys.stdout.flush()

mc.production()
progress.update(1)
sys.stdout.flush()

progress.close()

```

```python
loadings = mc.systems[0].averageLoadings
status = loadings.writeAveragesStatistics(mc.systems[0].components, mc.systems[0].frameworkMass())
print(status)
```

```python
pressure = mc.systems[0].inputPressure
print(f"The fugacity is: {pressure} Pa")

loading, loading_error = mc.systems[0].averageLoadings.averageLoading()
print(f"The absolute loading is: {loading.numberOfMolecules[0]} +/- {loading_error.numberOfMolecules[0]} molecules/cell")
print(f"The excess loading is: {loading.numberOfMolecules[0] - mc.systems[0].components[0].amountOfExcessMolecules} +/- {loading_error.numberOfMolecules[0]} molecules/cell")
```
