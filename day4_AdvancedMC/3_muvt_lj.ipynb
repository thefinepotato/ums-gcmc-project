# Exercise 3: Lennard-Jones in the Grand-Canonical ensemble

## Introduction

In the **grand-canonical ensemble** (GCE), a system of particles is described by constant volume $V$, temperature $T$, and chemical potential $\mu$, but the **number of particles** $N$ is **allowed to fluctuate**. This makes the GCE especially useful for studying processes where particles can be exchanged with a reservoir, such as adsorption, reactions, or phase equilibria. Physically, the system is in contact with both a **thermal reservoir** (fixing $T$) and a **particle reservoir** (fixing $\mu$). The probability of observing a configuration with $N$ particles and internal energy $U$ is then proportional to  
$
\exp\!\bigl[-\beta (U - \mu N)\bigr],
$
where $\beta = 1 / (k_B T)$.

To sample this ensemble using Monte Carlo (MC) techniques, one augments standard moves (e.g., particle displacements) with **insertions** and **deletions** of particles:

- **Insertion**: Attempt to add a new particle at a random position (and orientation if necessary).  
- **Deletion**: Attempt to remove an existing particle at random.

These moves enable changes in $N$, thereby reflecting the actual GCE dynamics of particle exchange.

---

## Physical and Theoretical Foundations

1. **Grand-Canonical Probability**  
   In the GCE, the distribution of states is governed by the factor $\exp[-\beta(U - \mu N)]$. Allowing $N$ to fluctuate provides access to phenomena like phase equilibria and density variations under changing chemical potential.

2. **Detailed Balance**  
   Because the number of particles can change, each insertion or deletion move must satisfy the **detailed balance** condition so that the Markov chain samples from the correct GCE distribution. This leads directly to **acceptance criteria** that incorporate both $\Delta U$ (the change in internal energy) and $\mu$ (the chemical potential).

---

## Mathematical Description and Acceptance Criteria

Let $\Delta U$ represent the energy change associated with adding or removing a particle. The acceptance probabilities for insertion and deletion moves ensure the simulation remains faithful to the grand-canonical ensemble.

### Insertion Move $(N \to N + 1)$

1. **Proposal**: Select a position uniformly in the volume $V$ (and orientation, if relevant) to place a new particle.  
2. **Energy Cost**: Calculate $\Delta U = U_{N+1} - U_N$.  
3. **Acceptance**: Accept the move with probability  
$
p_{\text{acc}}(N \to N+1) \;=\; \min \!\Bigl(1, \,
  \frac{\beta V}{N+1} \exp\!\bigl[-\beta (\Delta U - \mu)\bigr]
\Bigr).
$

### Deletion Move $(N \to N - 1)$

1. **Proposal**: Select one of the $N$ existing particles at random to remove.  
2. **Energy Cost**: Calculate $\Delta U = U_{N-1} - U_N$.  
3. **Acceptance**: Accept the move with probability  
$
p_{\text{acc}}(N \to N-1) \;=\; \min \!\Bigl(1, \,
  \frac{N}{\beta V} \exp\!\bigl[-\beta (\Delta U + \mu)\bigr]
\Bigr).
$

In each case, the factors $\frac{\beta V}{N+1}$ or $\frac{N}{\beta V}$ emerge from balancing the probabilities for forward (insertion) and reverse (deletion) moves, ensuring that the grand-canonical distribution is sampled correctly. The exponential terms account for the Boltzmann factor associated with either adding or removing a particle, including the shift by $\mu$.
