# Exercise 4: $NPT$ Lennard-Jones
Here, we will learn how to implement Monte Carlo simulations in the isothermal-isobaric ensemble. To do this we will make use of the Monte Carlo Lennard-Jones code introduced in the Basic Monte Carlo tutorial. First, we will have to implement a Monte Carlo move that alters the box size. The partition function of the isothermal-isobaric ensemble can be derived from statistical mechanical principles (p.184 Frenkel & Smit) and is defined as:

\begin{equation}
Q(N, P, T) = \frac{\beta P}{\Lambda^{3N}N!} \int d V V^{N} \exp(-\beta PV) \int d s^{N} \exp(-\beta \mathcal{U}(s^{N}; L))
\end{equation}

from which the probability density of changing the probability from $V$ to $V'$ can be derived. This yields the acceptance probability

\begin{equation}
\mathrm{acc}(o \to n) = \min \left(1, \exp \left[ -\beta \left(\mathcal{U}(s^N, V') - \mathcal{U}(s^N, V) + P(V'-V) - \frac{N}{\beta} \log \left( \frac{V'}{V} \right) \right) \right] \right)
\end{equation}

which is composed of the difference in the potential energy $\Delta \mathcal{U}$, the work done by the expansion $P \Delta V$ and an entropic contribution.

You will find the volume move in `molsim/monteCarlo/moves.cpp` in the method `MonteCarlo::volumeMove`. There are two phase, first the construction phase, where a perturbation is proposed and then an acceptance phase. In the construction phase you will find the random sampling of a perturbation magnitude (`volumeChange`) and a set of scaled coordinates `trialPositions`. To determine the energy difference a full computation of the energy needs to be carried out, as all interatomic distances are scaled.
